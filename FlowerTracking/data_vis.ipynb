# View flower-tracking data

Inspect recorded flower and moth positions from the included tracking tensors.
Choose a filename in the final cell to view another recording condition.

```python
from pathlib import Path
import os
import sys

PACKAGE_ROOT = next(
    (p for p in [Path.cwd(), *Path.cwd().parents]
     if (p / "NectarySearch").is_dir() and (p / "FlowerTracking").is_dir()),
    None,
)
if PACKAGE_ROOT is None:
    raise RuntimeError("Open this notebook from the extracted code folder.")
NOTEBOOK_DIR = PACKAGE_ROOT / "FlowerTracking"
sys.path.insert(0, str(PACKAGE_ROOT))
sys.path.insert(0, str(NOTEBOOK_DIR))
os.chdir(NOTEBOOK_DIR)

import numpy as np
import matplotlib.pyplot as plt
from task import MothDataset

plt.rcParams['figure.figsize'] = (20, 5)
data_dir = "data"
```

```python
nu = 0.06
sigma = 0.01

dataset = MothDataset(data_dir, nu=nu, sigma=sigma, normalize=False, seq_len=1000)
```

```python
files = dataset.seq_files
files = list(np.unique(files))
files
```

```python
indcs = np.argwhere(np.array(dataset.seq_files) == files[-1]).squeeze()

for i in indcs:

    st1, at1, xt1 = dataset.__getitem__(i)

    plt.figure()
    plt.plot(st1.numpy(), label='flower')
    plt.plot(at1.numpy(), label='moth')
    plt.legend()
    plt.title(dataset.seq_files[i])
```
